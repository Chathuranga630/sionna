```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Amal", "Nimal", "Kamal", "Saman"],
    "Age": [25, np.nan, 30, 28],
    "Salary": [50000, 60000, np.nan, 70000]
}

employees = pd.DataFrame(data)

print("Original data:")
display(employees)


# Identify missing values
print("Missing values in each column:")
print(employees.isnull().sum())


# Replace missing age with the average age
employees["Age"] = employees["Age"].fillna(employees["Age"].mean())


# Replace missing salary with the average salary
employees["Salary"] = employees["Salary"].fillna(
    employees["Salary"].mean()
)


print("\nData after handling missing values:")
display(employees)


employees.to_csv("emp.csv",index=False)
```

```python
import matplotlib.pyplot as plt
import numpy as np

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
monthly_sales = [120, 150, 140, 190, 220, 250]

products = ["Laptop", "Mobile", "Tablet", "Monitor"]
product_sales = [60, 95, 45, 70]

advertising = [10, 15, 20, 25, 30, 35]
sales_results = [110, 135, 150, 185, 215, 240]

customer_ratings = [2, 3, 3, 4, 4, 4, 5, 5, 5, 5]

# Create a 2 x 2 dashboard
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Dashboard title
fig.suptitle(
    "Business Sales Dashboard",
    fontsize=18,
    fontweight="bold"
)

# 1. Line chart
axes[0, 0].plot(
    months,
    monthly_sales,
    marker="o",
    color="blue",
    linewidth=2
)

axes[0, 0].set_title("Monthly Sales Trend")
axes[0, 0].set_ylabel("Sales (LKR '000)")
axes[0, 0].grid(True, alpha=0.3)

# 2. Horizontal bar chart
axes[0, 1].barh(
    products,
    product_sales,
    color="seagreen"
)

axes[0, 1].set_title("Product Sales")
axes[0, 1].set_xlabel("Units Sold")

# 3. Scatter plot with trend line
axes[1, 0].scatter(
    advertising,
    sales_results,
    color="orange",
    s=80
)

trend = np.polyfit(advertising, sales_results, 1)
trend_line = np.poly1d(trend)

axes[1, 0].plot(
    advertising,
    trend_line(advertising),
    color="red",
    linestyle="--",
    label="Trend line"
)

axes[1, 0].set_title("Advertising vs Sales")
axes[1, 0].set_xlabel("Advertising Cost")
axes[1, 0].set_ylabel("Sales")
axes[1, 0].legend()

# 4. Histogram
axes[1, 1].hist(
    customer_ratings,
    bins=[1.5, 2.5, 3.5, 4.5, 5.5],
    color="purple",
    edgecolor="black"
)

axes[1, 1].set_title("Customer Rating Distribution")
axes[1, 1].set_xlabel("Rating")
axes[1, 1].set_ylabel("Number of Customers")
axes[1, 1].set_xticks([2, 3, 4, 5])

plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Advertising expense
X = np.array([10, 15, 20, 25, 30, 35, 40]).reshape(-1, 1)

# Sales amount
y = np.array([100, 125, 150, 170, 200, 225, 250])

#create the model
model = LinearRegression()

#Traing the model

model.fit(X,y)


# Make prediction
predicted_sales = model.predict(X)


# Evaluate the model
print("R-squared:", r2_score(y, predicted_sales))



# Predict sales for a new advertising expense
new_advertising = [[32]]
new_prediction = model.predict(new_advertising)

print("Predicted sales for advertising expense 45:")
print(round(new_prediction[0], 2))


# Display the results
plt.scatter(X, y, color="blue", label="Actual sales")
plt.plot(X, predicted_sales, color="red", label="Regression line")

plt.title("Advertising Expense vs Sales")
plt.xlabel("Advertising Expense")
plt.ylabel("Sales")
plt.legend()
plt.grid(True, alpha=0.3)

plt.show()
```
